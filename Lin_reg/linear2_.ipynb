```python
       
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt   
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import warnings
warnings.filterwarnings('ignore')
```

```python
df2=pd.read_csv(r"C:\Users\bhagy\OneDrive\Desktop\practice\50_Startups.csv")
df2.head()
```

```python
df2.isnull().sum()
```

```python
x=df2.iloc[:,:-1]
y=df2.iloc[:,-1]
x.head()
```

```python
from sklearn.preprocessing import LabelEncoder
x['State']=LabelEncoder().fit_transform(x['State'])
x.head()
```

```python
x.corr()
```

```python
from sklearn.preprocessing import Normalizer
X2=Normalizer().fit_transform(x)
X2
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(X2,y,test_size=0.1)
print(xtrain.shape)
print(xtest.shape)
```

```python
M2=LinearRegression().fit(xtrain,ytrain)
print(M2)
```

```python
predict=M2.predict(xtest)
predict
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
Mean_squared_error = mean_squared_error(ytest,predict)
print('Mean_squared_error:',Mean_squared_error)

mean_absolute_error = mean_absolute_error(ytest,predict)
print('Mean_absolute_error:',mean_absolute_error)

R_squared = r2_score(ytest,predict)
print('R_squared:',R_squared)
```
